```python
# Step1. NumPyをインポートする
import numpy as np

# Step2. AとBの行列を格納する多次元配列のコードを記述する
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

# Step3. 行列の積の計算をする
product_matrix = np.dot(A, B)

# Step4. Step3の計算結果を表示する
print("行列の積:")
print(product_matrix)

# Step5. 行列の要素の最大値を求める
max_value = np.max(product_matrix)

# 最大値の表示
print("行列の要素の最大値:", max_value)
```
